```python
X = df.drop("Gender",axis=1)
y = df["Gender"]
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler,LabelEncoder,OneHotEncoder
from sklearn.impute import SimpleImputer
```

```python
num_cols = X.select_dtypes(exclude="object").columns
cat_cols = X.select_dtypes(include="object").columns
```

```python
print(num_cols)
cat_cols
```

```python
num_trans = Pipeline(steps=[
    ("imputer",SimpleImputer(strategy="median")),
    ("Standar",StandardScaler())
])
cat_trans = Pipeline(steps=[
    ("encoder",OneHotEncoder(drop="first"))
])

preprocess = ColumnTransformer([
    ("StandardScaler",num_trans,num_cols),
    ("OneHotEncoder",cat_trans,cat_cols)
])
```

```python
x_train, x_test, y_train, y_test = train_test_split(X,y,test_size=0.25,random_state=42)
```

```python
le = LabelEncoder()
y_train_trans = le.fit_transform(y_train)
y_test_trans = le.transform(y_test)
```

```python
model = GradientBoostingClassifier(
    loss="log_loss",          # loss function
    learning_rate=0.05,       # shrinkage rate
    n_estimators=300,         # number of boosting stages
    subsample=0.8,            # fraction of samples per tree
    max_depth=5,              # depth of individual trees
    min_samples_split=5,      # minimum samples to split a node
    min_samples_leaf=2,       # minimum samples per leaf
    max_features="sqrt",      # features considered at each split
    random_state=42 
)
model_train = Pipeline([
    ("preprocess",preprocess),
    ("model",model)
])
```

```python
model_train.fit(x_train,y_train_trans)
```

```python
pred = model_train.predict(x_test)
```

```python
print(pred[2])
print(y_test_trans[2])
```

```python
from sklearn.metrics import accuracy_score,roc_auc_score,classification_report,confusion_matrix,r2_score

print(r2_score(y_test_trans,pred))
```

```python
print("Accuracy:", accuracy_score(y_test_trans, pred))
```

```python
from sklearn.metrics import roc_curve, roc_auc_score
import matplotlib.pyplot as plt

# Predicted probability for the positive class
y_prob = model_train.predict_proba(x_test)[:, 1]

# Calculate ROC curve
fpr, tpr, thresholds = roc_curve(y_test_trans, y_prob)

# Calculate AUC
auc = roc_auc_score(y_test_trans, y_prob)

# Plot ROC curve
plt.figure(figsize=(8, 6))

plt.plot(fpr, tpr, label=f"Gradient Boosting (AUC = {auc:.3f})")

# Random classifier / no-skill line
plt.plot([0, 1], [0, 1], linestyle="--", label="Random Classifier")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.legend()
plt.grid()

plt.show()
```
