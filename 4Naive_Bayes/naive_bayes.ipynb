```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

## Loading dataset

```python

data = sns.load_dataset('tips')
```

```python
#Checking the uniques classes in each columns

print(data["time"].unique())
print(data["smoker"].unique())
print(data["sex"].unique())
print(data["day"].unique())
```

# Encoding the categorical valuues

```python
from sklearn.preprocessing import LabelEncoder,OneHotEncoder

label = LabelEncoder()
onehot = OneHotEncoder()
```

```python
# label Encoding
data["sex"] = label.fit_transform(data["sex"])
data["smoker"] = label.fit_transform(data["smoker"])
data["time"] = label.fit_transform(data["time"])
```

```python
# One-Hot Encoding
data = pd.get_dummies(data,columns=["day"],dtype=int)
```

```python
data.columns
```

```python
data.head()
```

```python
X = data.drop(columns="smoker")
y = data["smoker"]
```

# Splitting data into train-test

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.25,random_state=42)
```

```python
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

# Model Training

```python
from sklearn.naive_bayes import GaussianNB

gnb = GaussianNB()

gnb.fit(X_train,y_train)

pred = gnb.predict(X_test)
```

```python
# Metrices 

from sklearn.metrics import accuracy_score,confusion_matrix,classification_report

print(f"Accuracy: {accuracy_score(y_test, pred) * 100:.2f}%")
print("confusion matrix = \n", confusion_matrix(y_test,pred))
print("\nclassification report = \n", classification_report(y_test,pred))
```
